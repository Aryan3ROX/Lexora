```python
import torch
from sentence_transformers import SentenceTransformer
from sentence_transformers.util import cos_sim
```

```python
# Load book embedder
data = torch.load("artifacts/embedder/book_embeddings.pt", map_location="cpu")
emb = data["embeddings"]
meta = data["meta"]
titles = meta["titles"]

fine_tuned_model = SentenceTransformer('./artifacts/fine_tuned_book_embedder')
```

```python
# user_query = "fantasy, magical school, wizards, students"
user_query = "corrupt, politics"
query_embedding = fine_tuned_model.encode(user_query, convert_to_tensor=True).cpu()

# Find the 20 most similar books using cosine similarity
cosine_scores = cos_sim(query_embedding, emb)
top_results = torch.topk(cosine_scores[0], k=20)

print(f"\nTop 20 results for the query: '{user_query}'")
indices = top_results.indices.cpu().tolist()
scores = top_results.values.cpu().tolist()
for score, idx in zip(scores, indices):
    print(f"- {titles[idx]} (Score: {score:.4f})")
```
